```python
df=df.drop_duplicates()
```

```python
df["rating"] = pd.to_numeric(df["rating"],errors="coerce")
```

```python
print(df["rating"].isnull().sum())
```

```python
print(df["rating"].head(10))
```

```python
df["cost"]=(
    df["cost"].astype(str).str.replace(r"[^0-9.]","",regex=True)
)
```

```python
df["cost"]=pd.to_numeric(df["cost"],errors="coerce")
```

```python
print(df["cost"].isnull().sum())
```

```python
print(df["cost"].head(10))
```

```python
df["rating_count_num"] = df["rating_count"].str.replace("+ ratings", "")
df["rating_count_num"] = df["rating_count_num"].replace("Too Few Ratings", None)
df["rating_count_num"] = df["rating_count_num"].str.replace("K", "000")
df["rating_count_num"] = pd.to_numeric(df["rating_count_num"])
```

```python
print(df[["rating_count","rating_count_num"]].head(20))
```

```python
print(df[["rating","cost","rating_count_num"]].dtypes)
```

```python
print(df[["rating","cost","rating_count_num"]].head(10))
```

```python
df["cost_band"] = pd.cut(df["cost"],bins=[0,200,400,float("inf")],labels=["Budget","Mid","Premium"],right=False)
```

```python
print(df["cost_band"].value_counts())
```

```python
df_cuisine = df.assign(cuisine=df["cuisine"].str.split(","))
```

```python
df_cuisine=df_cuisine.explode("cuisine")
```

```python
df_cuisine["cuisine"] = df_cuisine["cuisine"].str.strip()
```

```python
df_cuisine= df_cuisine.dropna(subset=["cuisine"])
```

```python
print(df_cuisine["cuisine"].isnull().sum())
```

```python
print(df_cuisine[["name","city","cuisine"]].head(20))
```

```python
cols = ["id","name","city","rating","rating_count","rating_count_num","cost","cost_band","cuisine"]
```

```python
df[cols].to_csv("../../data/swiggy_dataset/swiggy_clean.csv", index=False)
```

```python
df_cuisine[cols].to_csv("../../data/swiggy_dataset/swiggy_cuisine.csv", index=False)
```
